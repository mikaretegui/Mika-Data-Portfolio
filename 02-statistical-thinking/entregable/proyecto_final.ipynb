```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
```

## I: LIMPIEZA Y MERGE DEL DATASET

```python
# Usamos la funcion de read_parquet
data_host= pd.read_parquet('airbnb_madrid_host.parquet')
data_location= pd.read_parquet('airbnb_madrid_location.parquet')
data_property= pd.read_parquet('airbnb_madrid_property.parquet')
data_condition= pd.read_parquet('airbnb_madrid_conditions.parquet')
data_review= pd.read_parquet('airbnb_madrid_reviews.parquet')
```

```python
# Hacemos merge de los dataset
data_full = data_condition.merge(
        right=data_host,
        on='id'
    ).merge(
        right=data_location,
        on='id'
    ).merge(
        right=data_property,
        on='id'
    ).merge(
        right=data_review,
        on='id'
    )
```

## II: Filtro el 80% de viviendas con menor precio para excluir el 20% mas caras y que el análisis sea más representativo
## Después, hago una separación entre el 10% de viviendas con mejor precio vs el resto para el análisis 


```python
#Calculo el percentil 80 del valor del precio
p80 = data_full['price'].quantile(0.80)
p80
```

```python
#Filtro el dataframe completo
data_full_80 = data_full[data_full['price'] <= p80]
data_full_80
```

```python
#Ahora calculo el percentil 90 del 80% de viviendas con precio más bajo y creo la columna de top_10
p90 = data_full_80['price'].quantile(0.90)
data_full_80['top10'] = data_full_80['price'] >= p90
data_full_80
```

# 1. Elaborar un contexto del problema

## CONTEXTO PARA EL ENTREGABLE

Madrid es un mercado turístico muy activo, donde el precio de una vivienda está influido principalmente por el tipo de vivienda y su ubicación.

En nuestro análisis partimos de cuatro elementos que ayudan a entender el punto de partida del cliente:

Tipo de vivienda (property type): Madrid ofrece desde estudios hasta apartamentos amplios, y esta diferencia condiciona el precio y el tipo de huésped que atraen.

Barrio (neighbourhood): No todas las zonas de Madrid tienen la misma demanda ni el mismo nivel de actividad turística, lo que impacta directamente en el precio.

Cercanía a Sol (dist_km_sol): Estar cerca del centro histórico suele asociarse a mayor demanda turística y, por tanto, a precios más altos.

Cercanía al aeropuerto (dist_km_airport): Aunque con menos peso que la zona centro, la accesibilidad al aeropuerto favorece a viajeros de negocios o estancias cortas.

Estos factores nos permiten entender el entorno en el que compiten las viviendas de nuestro cliente. A partir de aquí, el análisis se centra en las acciones que podemos tomar para mejorar el precio.

# 3. Definir y desarrollar el Roadmap del analisis

## GRUPO 1: Analizamos la reputación del anfitrión y como puede variar en el precio

```python
# Etiquetas legibles
df["top10_str"] = df["top10"].map({True: "Top 10%", False: "Resto"})
df["superhost_str"] = df["host_is_superhost"].map({1: "Superhost", 0: "No superhost"})
df["identity_str"] = df["host_identity_verified"].map({1: "Identidad verificada",
                                                       0: "Sin verificación"})

# Limpieza mínima para que los gráficos no se rompan
df["review_scores_rating_clean"] = df["review_scores_rating"].fillna(-1)
df["host_response_rate_clean"] = df["host_response_rate"].fillna(0)

# Figura con 4 gráficos (2x2)
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# 1) Precio vs Superhost
sns.boxplot(data=df, x="superhost_str", y="price", ax=axs[0,0])
axs[0,0].set_title("Precio según si el anfitrión es Superhost")
axs[0,0].set_xlabel("Superhost")
axs[0,0].set_ylabel("Precio (€)")

# 2) Precio vs Identidad verificada
sns.boxplot(data=df, x="identity_str", y="price", ax=axs[0,1])
axs[0,1].set_title("Precio según identidad verificada")
axs[0,1].set_xlabel("Estado de verificación")
axs[0,1].set_ylabel("Precio (€)")

# 3) Rating vs Top10 vs resto (solo anuncios con rating >= 0)
sns.boxplot(data=df[df["review_scores_rating_clean"] >= 0],
            x="top10_str", y="review_scores_rating_clean", ax=axs[1,0])
axs[1,0].set_title("Valoración media según grupo de precio")
axs[1,0].set_xlabel("Grupo")
axs[1,0].set_ylabel("Rating medio")

# 4) Tasa de respuesta vs Top10 vs resto
sns.boxplot(data=df, x="top10_str", y="host_response_rate_clean", ax=axs[1,1])
axs[1,1].set_title("Tasa de respuesta del anfitrión según grupo de precio")
axs[1,1].set_xlabel("Grupo")
axs[1,1].set_ylabel("Tasa de respuesta (%)")

plt.tight_layout()
plt.show()
```

1. Ser Superhost está asociado a precios ligeramente más altos

Los Superhosts muestran una mediana de precio algo superior y menos presencia en los rangos bajos.
Aunque la diferencia no es enorme, sí es consistente. Convertirse en Superhost es una palanca realista y accionable para incrementar precio.

2. La verificación de identidad no influye en el precio

Las distribuciones de precio entre anfitriones verificados y no verificados son prácticamente idénticas. No parece ser un factor que el huésped valore a nivel de precio.

3. El rating medio tampoco diferencia entre alojamientos baratos y caros

Tanto el top 10% de viviendas como el resto mantienen valoraciones muy altas y muy similares.
Esto es común en Airbnb (sesgo positivo en las reviews). La puntuación media no sirve como elemento diferenciador de precio.

4. La tasa de respuesta sí diferencia al top 10% de precio

Los alojamientos más caros muestran tasas de respuesta más altas y más estables, mientras que el resto presenta mucha más variabilidad. Responder rápido y de forma consistente es una característica del alojamiento premium.

### Conclusión: 
Entre todas las variables de reputación, solo dos muestran impacto real en el precio: Ser Superhost y mantener una tasa de respuesta alta

### Recomendación: 
El cliente debería centrarse en mejorar la comunicación (tasa de respuesta) y trabajar para obtener el estatus de Superhost, ya que son las palancas que más contribuyen a aumentar el valor de la vivienda.

## GRUPO 2: ¿Cómo puede afectar la capacidad y distribución del espacio al precio de una vivienda?

```python
# Figura 2x2
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# 1) Precio vs accommodates
sns.scatterplot(data=df, x="accommodates", y="price", alpha=0.3, ax=axs[0,0])
axs[0,0].set_title("Relación entre precio y capacidad (accommodates)")
axs[0,0].set_xlabel("Capacidad")
axs[0,0].set_ylabel("Precio (€)")

# 2) Boxplot precio vs bedrooms
sns.boxplot(data=df, x="bedrooms", y="price", ax=axs[0,1])
axs[0,1].set_title("Precio según número de dormitorios")
axs[0,1].set_xlabel("Dormitorios")
axs[0,1].set_ylabel("Precio (€)")

# 3) Boxplot precio vs bathrooms
sns.boxplot(data=df, x="bathrooms", y="price", ax=axs[1,0])
axs[1,0].set_title("Precio según número de baños")
axs[1,0].set_xlabel("Baños")
axs[1,0].set_ylabel("Precio (€)")

# 4) Correlaciones simples (heatmap pequeño)
subset = df[["price", "accommodates", "bedrooms", "bathrooms"]]
corr = subset.corr()

sns.heatmap(corr, annot=True, cmap="Blues", ax=axs[1,1])
axs[1,1].set_title("Correlación entre variables de capacidad y precio")

plt.tight_layout()
plt.show()
```

1. Capacidad (accommodates)

Existe una relación clara entre la capacidad de la vivienda y su precio: Las viviendas para 1–2 personas concentran los precios más bajos (15–60 €). A partir de 4 huéspedes, los precios se distribuyen también en los rangos altos y desaparecen muchas opciones baratas. A partir de 8–9 huéspedes, la oferta disminuye en todos los rangos de precio.

Insight clave: Las viviendas para grupos medianos (4–6 personas) son las que tienen más presencia en los tramos altos de precio. Aumentar la capacidad, si es factible, es una acción clara para incrementar el precio.

2. Número de dormitorios

El número de dormitorios también influye en el precio: Las viviendas con 0 dormitorios (estudios, lofts, hoteles) no son necesariamente más baratas que las de 1 dormitorio. El mayor salto de precio se observa al pasar de 1 a 2 dormitorios. A partir de 3–5 dormitorios, los precios se mantienen en rangos medios-altos de forma consistente. Para 6–7 dormitorios, la muestra es pequeña y no permite conclusiones firmes.

Insight clave: El salto más relevante ocurre al pasar de 1 a 2 dormitorios. Convertir espacios para añadir un segundo dormitorio puede ser una mejora muy rentable.

3. Número de baños

El número de baños muestra una dispersión elevada y no presenta un patrón claro: Las medianas son muy similares entre categorías. La variabilidad es alta. No se observan diferencias destacables entre grupos.

Insight clave: El número de baños no parece influir directamente en el precio.

4. Correlaciones entre las variables

El mapa de calor confirma lo comentado anteriormente: 

Capacidad ↔ precio: 0.56 (correlación más alta). 

Dormitorios ↔ precio: 0.30

Baños ↔ precio: prácticamente 0

Además, la capacidad está muy correlacionada con el número de dormitorios (0.57), lo cual es lógico.

Insight clave: La variable con mayor relación con el precio es la capacidad de la vivienda, seguida por el número de dormitorios. Los baños no muestran impacto relevante.

### Conclusión: 
El precio aumenta principalmente cuando la vivienda puede alojar a más personas, especialmente a partir de 4 huéspedes.
El número de dormitorios también contribuye, sobre todo al pasar de 1 a 2. El número de baños no se comporta como un factor determinante del precio.

### Recomendación: 
Priorizar mejoras que permitan aumentar la capacidad o añadir un dormitorio en la vivienda, ya que son las mejoras espaciales que realmente permiten acceder a tramos superiores de precio.

```python
# Etiquetas legibles
df["kitchen_str"] = df["amenities_kitchen"].map({1: "Tiene cocina", 0: "Sin cocina"})
df["tv_str"] = df["amenities_tv"].map({1: "Tiene TV", 0: "Sin TV"})
df["bedtype_str"] = df["bed_type"]

# Filtramos estancias mínimas razonables (<= 30 noches)
df_min_stay = df[df["minimum_nights"] <= 30].copy()

# Figura 2x2
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# (1) Precio según cocina
sns.boxplot(data=df, x="kitchen_str", y="price", ax=axs[0,0])
axs[0,0].set_title("Precio según si la vivienda tiene cocina")
axs[0,0].set_xlabel("Cocina")
axs[0,0].set_ylabel("Precio (€)")

# (2) Precio según TV
sns.boxplot(data=df, x="tv_str", y="price", ax=axs[0,1])
axs[0,1].set_title("Precio según si la vivienda tiene TV")
axs[0,1].set_xlabel("TV")
axs[0,1].set_ylabel("Precio (€)")

# (3) Precio vs estancia mínima (solo <= 30 noches)
sns.scatterplot(data=df_min_stay, x="minimum_nights", y="price", alpha=0.4, ax=axs[1,0])
axs[1,0].set_title("Relación entre precio y estancia mínima (<= 30 noches)")
axs[1,0].set_xlabel("Estancia mínima (noches)")
axs[1,0].set_ylabel("Precio (€)")

# (4) Precio según tipo de cama
sns.boxplot(data=df, x="bedtype_str", y="price", ax=axs[1,1])
axs[1,1].set_title("Precio según tipo de cama")
axs[1,1].set_xlabel("Tipo de cama")
axs[1,1].set_ylabel("Precio (€)")
axs[1,1].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

1. Cocina

Los alojamientos con cocina muestran precios claramente más altos que los que no la tienen. Las viviendas sin cocina incluyen sobre todo habitaciones privadas, hostels o alojamientos peculiares, lo que explica la presencia de outliers.

Insight: Incluir cocina es una de las amenidades más relevantes para aumentar el precio, especialmente en apartamentos y viviendas completas.

2. Televisión

Los alojamientos que incluyen TV también presentan precios más altos. De nuevo, los outliers en el grupo “sin TV” proceden de tipos de alojamiento no comparables.

Insight: La TV sigue siendo una amenidad básica muy valorada. Su presencia incrementa la disposición a pagar del huésped.

3. Estancia mínima (<= 30 noches)

No existe una relación clara entre la estancia mínima y el precio. Las estancias cortas (1–5 noches) muestran amplia variedad de precios; las estancias largas aparecen sobre todo en alojamientos más baratos, generalmente orientados a media-larga estancia.

Insight: La estancia mínima no explica diferencias de precio y no es un insight útil para el cliente. Se descarta como factor accionable.

4. Tipo de cama

Las diferencias de precio por tipo de cama no representan un efecto directo. Los futones aparecen en alojamientos más baratos; los sofás o airbeds, en alojamientos grandes donde funcionan como camas adicionales. Por tanto, el tipo de cama refleja la tipología y tamaño de la vivienda, no un impacto propio.

Insight: El tipo de cama no es una palanca para aumentar el precio. Solo está asociado a la capacidad o tamaño del alojamiento.

### Conclusión 
Las amenidades esenciales que sí influyen en el precio son la cocina y la TV. La estancia mínima y el tipo de cama no son factores accionables para incrementar el valor de la vivienda.

Priorizar amenities que aportan sensación de comodidad y hogar, especialmente cocina y televisión.
