### Definiciones de columnas del dataset

| Columna        | Descripción                                                                 |
|----------------|-----------------------------------------------------------------------------|
| invoice_no     | Número de factura. Una combinación de la letra 'I' y un número de 6 dígitos que identifica cada operación. |

# 11. ¿Cuál es el día de la semana con más ventas registradas en cada centro comercial?
Guarda el resultado en una variable llamada `dia_semana_mas_ventas_por_centro`. 

El tipo de `dia_semana_mas_ventas_por_centro` debe ser `pandas.core.series.Series` donde las etiquetas deben ser el nombre de los centros comerciales y los valores deben ser el nombre del dia de la semana con mas ventas.

```python
# El enunciado no especifica si “ventas” se refiere al número de transacciones, productos vendidos o al importe total generado.
# Por ese motivo, he calculado las tres métricas
```

```python
# Lo primero que hago es añadir la columna de días de la semana al DataFrame y compruebo que aparece
datos['dia_semana'] = datos['invoice_date'].dt.day_name()
datos
```

```python
# Primero calculo en base al número de facturas registradas en cada mall cada día de la semana
ventas_dia_centro_comercial = utils.ventas_mall_dia_semana(datos)
dia_semana_mas_ventas_por_centro = ventas_dia_centro_comercial.groupby('shopping_mall').idxmax()

dia_semana_mas_ventas_por_centro = dia_semana_mas_ventas_por_centro.apply(lambda x: x[1])
dia_semana_mas_ventas_por_centro
```

```python
# Compruebo que el tipo de la variable es pandas.core.series.Series
type(dia_semana_mas_ventas_por_centro)
```

```python
# Después agrupo en una variable por centro comercial y día de la semana la cantidad de productos vendidos
ventas_cantidad_centro_comercial = utils.cantidad_mall_dia_semana(datos)
dia_semana_mas_cantidad_por_centro = ventas_cantidad_centro_comercial.groupby('shopping_mall').idxmax()

dia_semana_mas_cantidad_por_centro = dia_semana_mas_cantidad_por_centro.apply(lambda x: x[1])
dia_semana_mas_cantidad_por_centro
```

```python
# Compruebo que el tipo de la variable es pandas.core.series.Series
type(dia_semana_mas_ventas_por_centro)
```

```python
# Lo podemos hacer también por los ingresos recibidos
ingresos_por_dia_centro_comercial = utils.ingresos_mall_dia_semana(datos)
dia_semana_mas_ingresos_por_centro = ingresos_por_dia_centro_comercial.groupby('shopping_mall').idxmax()

dia_semana_mas_ingresos_por_centro = dia_semana_mas_ingresos_por_centro.apply(lambda x: x[1])
dia_semana_mas_ingresos_por_centro
```

```python
# Compruebo que el tipo de la variable es pandas.core.series.Series
type(dia_semana_mas_ingresos_por_centro)
```
